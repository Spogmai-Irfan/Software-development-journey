```python
import pandas as pd
```

# Pandas_Basics_code

```python
data = {
    "Name": ["TALHOOO", "Sara", "Ahmed", "Huda"],
    "Age": [20, 22, 21, 23],
    "Marks": [85, 90, 78, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
result1 = df[(df["Age"]>21) & (df["Marks"]>80)]
print(result1)
result2 = df[(df["Age"] > 20)  | (df["Marks"] > 88)]

print(result2)
```

```python
df["Passed"] = df["Marks"] >= 80
print(df)
```

```python
df = df.sort_values("Marks")
print(df)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```

```python
df = df.dropna()
```

```python
print(df.duplicated())
```

```python
df = df.drop_duplicates()
```

```python
df = df.sort_values("Marks", ascending=False)

print(df)
```

```python
data = {
    "Gender": ["Male", "Female", "Male", "Male", "Female"]
}

df = pd.DataFrame(data)

print(df["Gender"].value_counts())
```

```python
data = {
    "Department": ["IT", "IT", "HR", "HR", "AI"],
    "Salary": [50000, 60000, 40000, 45000, 70000]
}

df = pd.DataFrame(data)

result = df.groupby("Department")["Salary"].mean()

print(result)
```
