```python
import pandas as pd

data = {
    "Name": ["TALHOOO", "Sara", "Ahmed", "Huda"],
    "Age": [20, 22, 21, 23],
    "Marks": [85, 90, 78, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Age"] = df["Age"].astype(int)
print(df.dtypes)
```

```python
def grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 60:
        return "B"
    else:
        return "C"

df["Grade"] = df["Marks"].apply(grade)

print(df)
```
