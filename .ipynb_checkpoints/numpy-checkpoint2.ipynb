# CREATION ARRAYS

1D Array

```python
arr = np.array([10, 20, 30, 40, 50])
print (arr)
```

2D Array

```python
matrix = np.array([[1,2,3],[4,5,6]])
print(matrix)
```

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr.shape)
print(arr.size)
print(arr.ndim)
```

```python
print(np.zeros((2,3)))
```

```python
print(np.ones((2,4)))
```

```python
print(np.eye(4))
```

```python
print(np.arange(1,8))
```

```python
print(np.arange(0,30,3))
```

```python
print(np.linspace(0, 10, 5))
```

```python
arr = np.arange(1, 17)

new_arr = arr.reshape(4, 4)

print(new_arr)
```

```python
print(np.linspace(1, 6, 6))
```

```python
arr = np.array([5, 66, 567, 999])

print(arr[0])
print(arr[2])
```

```python
matrix = np.array([[1, 2], [5, 4]])

print(matrix[1, 0])
```

```python
arr = np.array([1, 0, 3, 40, 0])

print(arr[1:4])
```

```python
arr = np.array([10, 20, 30, 40, 50,70])

print(arr[1:5])
```

```python
a = np.array([1, 2, 3])

print(a + 5)
print(a * 2)
print(a - 1)
print(a / 2)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)
print(a * b)
```

```python
arr = np.array([1, 4, 9, 16])

print(np.sqrt(arr))
print(np.sum(arr))
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
```

```python
print(np.random.rand(7))
```

```python
print(np.random.randint(1, 10, 5))
```

```python
arr = np.array([10, 25, 30, 15, 40])

print(arr > 20)
```

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print(arr.sum(axis=0))
print(arr.sum(axis=1))
```
