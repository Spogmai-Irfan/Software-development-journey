```python
import numpy as np
marks = np.array([45, 67, 89, 23, 90, 56, 78, 34, 99, 12])
print("All marks:", marks)
first_five = marks[0:5]
print("First 5 marks:", first_five)
last_three = marks[-3:]
print("Last 3 marks:", last_three)

alternate = marks[::2]
print("Every alternate mark:", alternate)
reversed_marks = marks[::-1]
print("Reversed marks:", reversed_marks)

passed = marks[marks > 50]
print("Marks greater than 50:", passed)
failed = marks[marks < 40]
print("Marks less than 40:", failed)
marks[0] = 100
print("After updating first mark:", marks)

marks[1:4] = [50, 60, 70]
print("After updating marks 1 to 3:", marks)
students = np.array([
    [45, 67, 89],
    [23, 90, 56],
    [78, 34, 99]
])

print("Full 2D array:")
print(students)

print("First row:", students[0])

print("First column:", students[:, 0])
print("Element at row 1, col 2:", students[1, 2])
sub_matrix = students[0:2, 0:2]
print("Sub matrix:")
print(sub_matrix)
```
