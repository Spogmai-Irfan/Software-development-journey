```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])
b = np.array([1, 2, 3, 4, 5])

print("Array A:", a)
print("Array B:", b)

print("Addition:", a + b)

print("Subtraction:", a - b)

print("Multiplication:", a * b)
print("Division:", a / b)

print("Sum:", np.sum(a))
print("Average:", np.mean(a))
print("Maximum:", np.max(a))
print("Minimum:", np.min(a))

matrix = np.arange(1, 10).reshape(3, 3)

print("3x3 Matrix:")
print(matrix)

print("Transpose:")
print(matrix.T)


```
