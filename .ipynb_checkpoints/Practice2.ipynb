```python
import numpy as np
```

```python
a= np.array([3,4,5,6,3])
print(np.sum(a))
print(np.mean(a))
print(np.max(a))
print(np.min(a))
print(np.std(a))
```

```python
A = np.array([[1,2],[4,6]])
B = np.array([[2,3],[4,6]])
print(A+B)
print(A@B)
print(np.transpose(A))
```

```python
a =np.array([34,53,56,44,5])
print(a[a>25])
```

```python
a=np.array([2,3,5,6,7,3])
print(np.sqrt(a))
print(np.sin(a))
print(np.log(a))
```

```python
a =np.array([1,2,3,4])
print(a+10)
```

```python
a=np.array([2,5,6,7])
print(np.sort(a))
```

```python
a =np.array([1,2,2,3,3,4])
print(np.unique(a))
```
