```python
X = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(X[0])       
print(X[:, 0])  

print(X[0, 1])    

print(X[0:2])     
print(X[:, 1:3])  
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
data = np.array([10, 20, 30, 40, 50])

print(np.mean(data))
print(np.sum(data))
print(np.min(data))
print(np.max(data))
```

```python
X = np.array([
    [1, 2],
    [3, 4]
])

W = np.array([
    [5, 6],
    [7, 8]
])

result = np.dot(X, W)

print(result)
```

```python
x = np.array([10, 20, 30, 40, 50])

print(x > 30)
```

```python
x = np.array([10, 20, 30, 40, 50])

x_normalized = (x - np.min(x)) / (np.max(x) - np.min(x))

print(x_normalized)
```

```python
x = np.array([10, 20, 30, 40, 50])

mean = np.mean(x)
std = np.std(x)

x_scaled = (x - mean) / std

print(x_scaled)
```

```python
import numpy as np
X = np.array([
    [1, 2],
    [2, 3],
    [3, 4],
    [4, 5]
])

W = np.array([
    [0.5],
    [0.8]
])

b = 0.1

Z = X @ W + b

print("Input:")
print(X)

print("Output:")
print(Z)
```
